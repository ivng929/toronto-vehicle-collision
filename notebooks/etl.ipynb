## 2. Transform

```python
# create a collision table to store information related to the collision
collision_columns = ["collision_id",
                     "accdate",
                     "stname1",
                     "stname2",
                     "stname3",
                     "per_inv",
                     "acclass",
                     "accloc",
                     "traffictl",
                     "impactype",
                     "visible",
                     "light",
                     "rdsfcond",
                     "road_class",
                     "failtorem",
                     "longitude",
                     "latitude",
                     "wardname",
                     "division",
                     "neighbourhood",
                     "aggressive",
                     "distracted",
                     "cyclist",
                     "motorcyclist",
                     "other_micromobility",
                     "older_adult",
                     "pedestrian",
                     "red_light",
                     "school_child",
                     "heavy_truck"]

collision_df = (df[collision_columns].drop_duplicates())
```

```python
collision_df.head()
```

```python
collision_df.shape
```

```python
# create a person table that stores information about
# each person involved in each collision
person_columns = ["collision_id",
                  "per_no",
                  "veh_no",
                  "vehtype",
                  "initdir",
                  "road_user",
                  "invage",
                  "injury",
                  "safequip",
                  "drivact",
                  "drivcond",
                  "pedact",
                  "pedcond",
                  "manoeuvre",
                  "pedtype",
                  "cyclistype",
                  "cycact",
                  "cyccond",
                  "fatal_no"]

person_df = df[person_columns].copy()

# add person_id column
person_df.insert(0, "persion_id", range(1, len(person_df)+1))

person_df.head()
```

```python
person_df.shape
```

## 3. Load

```python
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# read credentials
from urllib.parse import quote_plus
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = quote_plus(os.getenv("DB_PASSWORD"))
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")
DB_NAME = os.getenv("DB_NAME")
```

```python
# connect to postgres
from sqlalchemy import create_engine, text
engine = create_engine(f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}")
```

```python
# create schema
with open("../databases/schema.sql", "r") as file:
    sql_script = file.read()
with engine.begin() as conn:
    conn.execute(text(sql_script))
```

```python
# load collision table into postgres
collision_df.to_sql("collision", 
          engine, 
          if_exists="append",
          index=False,
          chunksize=20000,
          method="multi")
```

```python
# load person table into postgres
person_df.to_sql("person", 
          engine, 
          if_exists="replace",
          index=False,
          chunksize=20000,
          method="multi")
```
